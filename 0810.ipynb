```python
x = 1
y = 6
w1 = 2
w2 = 2

#1
h = x*w1
pred = h*w2
err = (pred-y)**2
```

| що рахуємо | як                          | результат |
|---|-----------------------------|-----------|
| як прогноз впливає на помилку | $2 \cdot (4 - 6)$           | $-4$      |
| крутизна по $w_2$ | $-4 \cdot h = -4 \cdot 2$   | $-8$      |
| як $h$ впливає на помилку | $-4 \cdot w_2 = -4 \cdot 2$ | $-8$      |
| крутизна по $w_1$ | $-8 \cdot x = -8 \cdot 1$   | $-8$      |


```python
# w1
w1_new = w1 + 0.001
h_new = x * w1_new
pred_new = h_new * w2
err_new = (pred_new - y) ** 2

print("Крутизна w1:", (err_new - err) / 0.001)

# w2
w2_new = w2 + 0.001
pred_new = h * w2_new
err_new = (pred_new - y) ** 2

print("Крутизна w2:", (err_new - err) / 0.001)
```

```python
x = 1
y = 6
w1 = 0
w2 = 0

h = x * w1
pred = h * w2
err = (pred - y) ** 2

d_pred = 2 * (pred - y)
d_w2 = d_pred * h
d_h = d_pred * w2
d_w1 = d_h * x

print("Помилка:", err)
print("Крутизна по w1:", d_w1)
print("Крутизна по w2:", d_w2)

# 3. Крок навчання
lr = 0.05

w1 = w1 - lr * d_w1
w2 = w2 - lr * d_w2

# 4. Нова помилка
h = x * w1
pred = h * w2
new_err = (pred - y) ** 2

print("Новий w1:", w1)
print("Новий w2:", w2)
print("Нова помилка:", new_err)
```
